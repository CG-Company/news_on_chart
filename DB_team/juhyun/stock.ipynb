```python
import psycopg2

# DB 연결
conn = psycopg2.connect(
    host="localhost",
    dbname="news_on_chart",
    user="postgres",
    password="0912"
)
cur = conn.cursor()

# 쿼리 실행
cur.execute("""
    SELECT *
    FROM stock_price
    WHERE ticker = '000660'
    ORDER BY price_date DESC
""")
rows = cur.fetchall()
for row in rows:
    print(row)

conn.close()
```
